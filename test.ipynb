3. **Value**: The exact structural cell content located at the intersection of the Entity and the Attribute.

### SYSTEM CONSTRAINTS & OPERATIONAL LAWS:
- **Strict Data Integrity**: Preserve all original text data exactly as it exists in the input cells. Do NOT modify casing, do NOT clean up spacing, do NOT remove symbols (like *, ±, -), and do NOT truncate units. Copy poorly formatted strings exactly as-is.
