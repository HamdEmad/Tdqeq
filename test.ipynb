```python
def unpack_json_to_output(data):
    output = []
    
    for table in data:
        # 1. Grab the original top-level fields
        page_number = table.get("page_number")
        confidence_score = table.get("confidence_score")
        table_title = table.get("table_title")
        column_mapping = table.get("column_mapping", {})
        
        unpacked_rows = []
        
        for row in table.get("data", []):
            # 2. Extract base/shared attributes (any key starting with 'c')
            base_attrs = {k: v for k, v in row.items() if k.startswith("c")}
            
            v_cols = row.get("v_cols", [])
            v_data = row.get("v", [])
            
            if v_data:
                # 3. If variants exist, merge each variant with the base attributes
                for variant_row in v_data:
                    combined = base_attrs.copy()
                    # Map the list of values to their corresponding 'cX' keys
                    for i, col_key in enumerate(v_cols):
                        combined[col_key] = variant_row[i]
                    
                    # 4. Translate 'cX' keys to human-readable names using column_mapping
                    translated_row = {
                        column_mapping.get(k, k): v for k, v in combined.items()
                    }
                    unpacked_rows.append(translated_row)
            else:
                # 5. If no variants exist, just translate the base attributes
                translated_row = {
                    column_mapping.get(k, k): v for k, v in base_attrs.items()
                }
                unpacked_rows.append(translated_row)
        
        # 6. Assemble the final object structure for this table
        output.append({
            "page_number": page_number,
            "confidence_score": confidence_score,
            "table_title": table_title,
            "rows": unpacked_rows
        })
                
    return output
```

```python
def unpack_optimized_to_eav(llm_output, primary_entity_id="c1"):
    """
    Converts the token-optimized nested LLM JSON back into a flat EAV list.
    
    :param llm_output: The JSON array output from the LLM.
    :param primary_entity_id: The dictionary key representing the Primary Entity (usually "c1").
    :return: A list of strict EAV dictionaries.
    """
    unpacked_eav_data = []

    # Iterate over each extracted table in the payload
    for table in llm_output:
        mapping = table.get("column_mapping", {})
        data = table.get("data", [])
        
        # Resolve the actual name of the Primary Entity (e.g., "Part No.")
        entity_header = mapping.get(primary_entity_id, "Primary_ID")
        
        for record in data:
            # Get the actual ID value for this row (e.g., "ESR01")
            entity_id = record.get(primary_entity_id, "Unknown")
            
            # Step 1: Unpack Shared Attributes (The root-level 'c_' keys)
            for key, value in record.items():
                if key.startswith("c") and key != primary_entity_id:
                    attribute_name = mapping.get(key, key) # Lookup real column name
                    unpacked_eav_data.append({
                        "Entity_Header": entity_header,
                        "Entity_ID": entity_id,
                        "Attribute": attribute_name,
                        "Value": value
                    })
            
            # Step 2: Unpack Variant Attributes (The 1-to-many matrix)
            v_cols = record.get("v_cols", [])
            v_matrix = record.get("v", [])
            
            for row in v_matrix:
                for col_index, cell_value in enumerate(row):
                    col_id = v_cols[col_index]
                    attribute_name = mapping.get(col_id, col_id) # Lookup real column name
                    
                    unpacked_eav_data.append({
                        "Entity_Header": entity_header,
                        "Entity_ID": entity_id,
                        "Attribute": attribute_name,
                        "Value": cell_value
                    })

    return unpacked_eav_data


# --- EXAMPLE USAGE ---

# Assume `optimized_json_payload` is the dictionary returned by your LLM API
# flat_eav_dataset = unpack_optimized_to_eav(optimized_json_payload)
# print(json.dumps(flat_eav_dataset, indent=2))
```
